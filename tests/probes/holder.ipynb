```python
mem_a = ''
mem_b = ''
shm_a = ''
shm_b = ''
max_seconds = 50
```

```python
import os, time
secret = mem_a + mem_b  # kept in this process's memory
with open('/dev/shm/holder_secret', 'w') as f:
    f.write(shm_a + shm_b)
print('holder ready', flush=True)
start = time.time()
while not os.path.exists('/space/release.txt') and time.time() - start < max_seconds:
    time.sleep(0.5)
with open('/dev/shm/holder_secret') as f:
    intact = secret == mem_a + mem_b and f.read() == shm_a + shm_b
print(f'holder done intact={intact} waited={time.time() - start:.1f}s', flush=True)
assert intact, 'holder secrets changed'
```
