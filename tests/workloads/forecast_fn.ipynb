```python
region = 'jakarta'
cfg = {'months': 3, 'include': ['sales']}
fail = False
bad_output = False
```

```python
import nbsandbox, pandas as pd
with open('models/model.txt') as f:
    model = f.read().strip()
if fail:
    raise ValueError('asked to fail')
months = cfg['months']
df = pd.DataFrame({'month': range(1, months + 1), 'value': [m * 1.5 for m in range(1, months + 1)]})
print('model', model, 'region', region, 'exact params', nbsandbox.parameters(), flush=True)
```

```python
if bad_output:
    nbsandbox.output({'oops': True})
else:
    nbsandbox.output({'region': region, 'model': model, 'forecast': df['value'], 'table': df, 'echo': nbsandbox.parameters()})
```
