```python
region = 'all'
months = 12
```

```python
import os, pandas as pd
df = pd.read_csv('data/sales.csv')
if region != 'all':
    df = df[df['region'] == region]
df = df[df['month'] <= months]
summary = df.groupby('region', as_index=False)['value'].sum()
summary.to_csv(os.path.join(os.environ['NBSANDBOX_OUTPUT_DIR'], 'summary.csv'), index=False)
print(f'region={region} months={months} rows={len(df)}')
print(summary.to_string(index=False))
```
