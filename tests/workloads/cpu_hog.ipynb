```python
seconds = 40
```

```python
import multiprocessing as mp, os, time
def spin(deadline):
    x = 0
    while time.time() < deadline:
        x += 1
n = (os.cpu_count() or 2) * 2
deadline = time.time() + seconds
ps = [mp.Process(target=spin, args=(deadline,)) for _ in range(n)]
[p.start() for p in ps]
[p.join() for p in ps]
print(f'hogged with {n} processes')
```
