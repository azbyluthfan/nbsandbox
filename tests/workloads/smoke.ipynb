```python
region = 'default'
months = 3
```

```python
import os, pandas as pd
df = pd.DataFrame({'month': range(1, months + 1), 'value': [m * 10 for m in range(1, months + 1)]})
print('region:', region)
print(df)
```

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
out = os.environ['NBSANDBOX_OUTPUT_DIR']
df.plot(x='month', y='value').get_figure().savefig(os.path.join(out, 'chart.png'))
df.to_csv(os.path.join(out, 'summary.csv'), index=False)
print('artifacts written')
```
