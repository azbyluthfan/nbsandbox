# Sales report

An example notebook for `examples/`: summarise one region's sales, return the totals with `nbsandbox.output`, and write the monthly table as an artifact.

```python
# Parameters: the caller's values replace these.
region = "jakarta"
months = 3
```

```python
import os
import pandas as pd
import nbsandbox

sales = pd.read_csv("data/sales.csv")  # files are read relative to the space
mine = sales[sales["region"] == region].sort_values("month").tail(months)
if mine.empty:
    raise ValueError(f"no sales for region {region!r}")
print(mine.to_string(index=False))
```

```python
# A file in NBSANDBOX_OUTPUT_DIR becomes an artifact of the run.
mine.to_csv(os.path.join(os.environ["NBSANDBOX_OUTPUT_DIR"], "monthly.csv"), index=False)

# The value the caller gets back.
nbsandbox.output({
    "region": region,
    "months": int(len(mine)),
    "total": float(mine["amount"].sum()),
    "by_month": mine[["month", "amount"]],
})
```
