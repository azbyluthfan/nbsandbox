# An HTTP API from a notebook

Calls GitHub's public API through the run's fetcher ([connections.md](../../docs/connections.md#7-call-an-http-api)): the notebook names `github`, a connection its space binds; the fetcher adds the credential (a token, if the connection has one) and reaches the API through the egress gateway. The notebook never sees the token or the network.

Input: `repo` (owner/name). Output: the repository's stars and its latest releases.

```python
repo = "jupyter/notebook"
```

```python
import nbsandbox

gh = nbsandbox.http("github")          # a requests.Session; paths are relative to the connection's URL
gh.headers["Accept"] = "application/vnd.github+json"

info = gh.get(f"repos/{repo}").json()
releases = gh.get(f"repos/{repo}/releases", params={"per_page": 5}).json()
latest = [{"tag": r["tag_name"], "published": r["published_at"][:10]} for r in releases]
print(f"{info['full_name']}: {info['stargazers_count']} stars")
for r in latest:
    print(r["published"], r["tag"])
```

```python
nbsandbox.output({"repo": info["full_name"], "stars": info["stargazers_count"], "releases": latest})
```
