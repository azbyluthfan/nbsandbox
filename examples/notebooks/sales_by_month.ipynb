# Sales by month (streamed)

An example notebook for `examples/`: send one region's sales month by month with `nbsandbox.emit`, so a caller can draw the chart as the data arrives. Its contract declares a stream of items; the result's `output` is the list of them.

```python
# Parameters: the caller's values replace these.
region = "jakarta"
months = 6
```

```python
import time
import pandas as pd
import nbsandbox

sales = pd.read_csv("data/sales.csv")  # files are read relative to the space
mine = sales[sales["region"] == region].sort_values("month").tail(months)
if mine.empty:
    raise ValueError(f"no sales for region {region!r}")
```

```python
running = 0.0
for i, row in enumerate(mine.itertuples(), start=1):
    time.sleep(0.5)  # stands in for real work per month
    running += row.amount
    # One item per month: the caller gets it now, and the result is the list.
    nbsandbox.emit({"month": row.month, "amount": float(row.amount), "running_total": running}, total=len(mine))
    nbsandbox.progress(i / len(mine), f"{i} of {len(mine)} months")
```
